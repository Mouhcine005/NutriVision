```python
from pathlib import Path
import pandas as pd
import os

PROJECT_ROOT = Path.cwd().parent

FOOD11_DIR = PROJECT_ROOT / "data" / "raw" / "food11" / "archive"
USDA_DIR = PROJECT_ROOT / "data" / "raw" / "usda_foundation"
```

```python
splits = ["training", "validation", "evaluation"]

for split in splits:
    split_path = FOOD11_DIR / split

    print(f"\n{split.upper()}")

    total_images = 0

    for category in sorted(split_path.iterdir()):
        if category.is_dir():
            images = [
                f for f in category.iterdir()
                if f.suffix.lower() in [".jpg", ".jpeg", ".png"]
            ]

            count = len(images)
            total_images += count

            print(f"{category.name}: {count}")

    print(f"TOTAL: {total_images}")
```

```python
csv_files = list(USDA_DIR.glob("*.csv"))

for file in csv_files:
    df = pd.read_csv(file, nrows=5)

    print("\n" + "=" * 60)
    print("FILE:", file.name)
    print("COLUMNS:", df.columns.tolist())
    print("SAMPLE:")
    display(df.head(2))
```
