```python
import json
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
USDA_DIR = PROJECT_ROOT / "data" / "raw" / "usda_foundation"

# Find the JSON file
json_files = list(USDA_DIR.glob("*.json"))

print("JSON files found:")
for file in json_files:
    print(file.name, "-", round(file.stat().st_size / 1024, 2), "KB")
```

```python
# Load the first JSON file
json_path = json_files[0]

with open(json_path, "r", encoding="utf-8") as f:
    usda_data = json.load(f)

print("Python data type:", type(usda_data))

if isinstance(usda_data, dict):
    print("Top-level keys:", list(usda_data.keys()))
elif isinstance(usda_data, list):
    print("Number of records:", len(usda_data))
    if usda_data:
        print("First record keys:", list(usda_data[0].keys()))
```

```python
foods = usda_data["FoundationFoods"]

print("Number of food records:", len(foods))

# Inspect the first food
first_food = foods[0]

print("\nFood record fields:")
print(list(first_food.keys()))

print("\nFood ID:", first_food.get("fdcId"))
print("Description:", first_food.get("description"))
print("Data type:", first_food.get("dataType"))

print("\nNumber of nutrient entries:",
      len(first_food.get("foodNutrients", [])))

print("\nFirst 5 nutrient entries:")
for nutrient in first_food.get("foodNutrients", [])[:5]:
    print(nutrient)
```

```python
# Inspect important nutrients in the first food
target_nutrients = [
    "Energy",
    "Energy (Atwater General Factors)",
    "Protein",
    "Carbohydrate, by difference",
    "Total lipid (fat)"
]

for item in first_food["foodNutrients"]:
    nutrient = item.get("nutrient", {})
    name = nutrient.get("name")

    if name in target_nutrients:
        print({
            "name": name,
            "amount": item.get("amount"),
            "unit": nutrient.get("unitName"),
            "nutrient_id": nutrient.get("id")
        })
```

```python
all_nutrients = set()

for food in foods:
    for item in food.get("foodNutrients", []):
        nutrient = item.get("nutrient", {})
        if nutrient.get("name"):
            all_nutrients.add(nutrient["name"])

print("Unique nutrient names:", len(all_nutrients))

for name in sorted(all_nutrients):
    print(name)
```

```python
import pandas as pd

target_nutrient_ids = {
    1008: "calories_kcal",
    1003: "protein_g",
    1004: "fat_g",
    1005: "carbohydrates_g"
}

nutrition_records = []

for food in foods:
    for item in food.get("foodNutrients", []):

        nutrient = item.get("nutrient", {})
        nutrient_id = nutrient.get("id")

        if nutrient_id in target_nutrient_ids:

            nutrition_records.append({
                "fdc_id": food.get("fdcId"),
                "food_name": food.get("description"),
                "nutrient_name": target_nutrient_ids[nutrient_id],
                "amount": item.get("amount"),
                "unit": nutrient.get("unitName")
            })

nutrition_df = pd.DataFrame(nutrition_records)

display(nutrition_df.head(15))

print("Total extracted records:", len(nutrition_df))
print("Unique foods:", nutrition_df["fdc_id"].nunique())
print("\nNutrient distribution:")
print(nutrition_df["nutrient_name"].value_counts())
```

```python
energy_ids = [1008, 1062, 2047, 2048]

energy_records = []

for food in foods:
    for item in food.get("foodNutrients", []):
        nutrient = item.get("nutrient", {})

        if nutrient.get("id") in energy_ids:
            energy_records.append({
                "fdc_id": food.get("fdcId"),
                "food_name": food.get("description"),
                "nutrient_id": nutrient.get("id"),
                "nutrient_name": nutrient.get("name"),
                "amount": item.get("amount"),
                "unit": nutrient.get("unitName")
            })

energy_df = pd.DataFrame(energy_records)

display(energy_df.head(20))

print("\nEnergy measurement counts:")
print(energy_df.groupby(
    ["nutrient_id", "nutrient_name", "unit"]
).size())

print("\nUnique foods per energy measurement:")
print(energy_df.groupby("nutrient_id")["fdc_id"].nunique())
```

```python
energy_coverage = {}

for nutrient_id in [1008, 2047, 2048]:
    energy_coverage[nutrient_id] = set(
        energy_df.loc[
            energy_df["nutrient_id"] == nutrient_id,
            "fdc_id"
        ]
    )

standard = energy_coverage[1008]
general = energy_coverage[2047]
specific = energy_coverage[2048]

print("Standard kcal:", len(standard))
print("General Atwater:", len(general))
print("Specific Atwater:", len(specific))

print("\nFoods covered by standard kcal + specific Atwater:",
      len(standard | specific))

print("Foods covered by all three:",
      len(standard | general | specific))

print("\nFoods with no energy measurement:",
      424 - len(standard | general | specific))

print("\nSpecific Atwater only:",
      len(specific - standard))

print("General Atwater only:",
      len(general - standard - specific))
```

```python
# Check overlaps between energy measurements

print("Standard kcal AND Specific Atwater:",
      len(standard & specific))

print("Standard kcal AND General Atwater:",
      len(standard & general))

print("Specific Atwater AND General Atwater:",
      len(specific & general))

print("Foods with all three measurements:",
      len(standard & specific & general))

# Show examples of foods with multiple energy measurements
energy_comparison = energy_df.pivot_table(
    index=["fdc_id", "food_name"],
    columns="nutrient_id",
    values="amount",
    aggfunc="first"
).reset_index()

display(energy_comparison.head(15))
```

```python
# Energy priority: standard kcal > specific Atwater > general Atwater

energy_priority = [1008, 2048, 2047]

energy_rows = []

for food in foods:
    selected = None

    for nutrient_id in energy_priority:
        matches = [
            item for item in food.get("foodNutrients", [])
            if item.get("nutrient", {}).get("id") == nutrient_id
        ]

        if matches and matches[0].get("amount") is not None:
            selected = matches[0]
            break

    if selected:
        nutrient = selected["nutrient"]

        energy_rows.append({
            "fdc_id": food["fdcId"],
            "food_name": food["description"],
            "calories_kcal": selected["amount"],
            "energy_nutrient_id": nutrient["id"],
            "energy_method": nutrient["name"]
        })
    else:
        energy_rows.append({
            "fdc_id": food["fdcId"],
            "food_name": food["description"],
            "calories_kcal": None,
            "energy_nutrient_id": None,
            "energy_method": None
        })

energy_clean_df = pd.DataFrame(energy_rows)

display(energy_clean_df.head(15))

print("Total foods:", len(energy_clean_df))
print("Foods with energy:", energy_clean_df["calories_kcal"].notna().sum())
print("Foods missing energy:", energy_clean_df["calories_kcal"].isna().sum())

print("\nEnergy method distribution:")
print(energy_clean_df["energy_method"].value_counts(dropna=False))
```

```python
# Extract protein, fat and carbohydrates
target_ids = {
    1003: "protein_g",
    1004: "fat_g",
    1005: "carbohydrates_g"
}

macro_records = []

for food in foods:
    record = {
        "fdc_id": food["fdcId"],
        "food_name": food["description"]
    }

    for nutrient_id, column in target_ids.items():
        matches = [
            item for item in food.get("foodNutrients", [])
            if item.get("nutrient", {}).get("id") == nutrient_id
        ]

        record[column] = (
            matches[0].get("amount")
            if matches else None
        )

    macro_records.append(record)

macro_df = pd.DataFrame(macro_records)

# Combine macronutrients with selected energy values
nutrition_clean_df = macro_df.merge(
    energy_clean_df[
        ["fdc_id", "calories_kcal",
         "energy_nutrient_id", "energy_method"]
    ],
    on="fdc_id",
    how="left",
    validate="one_to_one"
)

# Arrange columns
nutrition_clean_df = nutrition_clean_df[
    [
        "fdc_id",
        "food_name",
        "calories_kcal",
        "protein_g",
        "fat_g",
        "carbohydrates_g",
        "energy_nutrient_id",
        "energy_method"
    ]
]

display(nutrition_clean_df.head(15))

print("\nDataset shape:", nutrition_clean_df.shape)

print("\nMissing values:")
print(nutrition_clean_df.isna().sum())

print("\nDuplicate FDC IDs:",
      nutrition_clean_df["fdc_id"].duplicated().sum())
```

```python
# Check negative nutritional values
nutrition_columns = [
    "calories_kcal",
    "protein_g",
    "fat_g",
    "carbohydrates_g"
]

print("Negative values:")
for column in nutrition_columns:
    count = (nutrition_clean_df[column] < 0).sum()
    print(f"{column}: {count}")

# Basic statistical summary
display(nutrition_clean_df[nutrition_columns].describe())

# Inspect foods with missing nutritional values
display(
    nutrition_clean_df[
        nutrition_clean_df[nutrition_columns].isna().any(axis=1)
    ].head(10)
)
```

```python
negative_carbs = nutrition_clean_df[
    nutrition_clean_df["carbohydrates_g"] < 0
].copy()

display(
    negative_carbs[
        ["fdc_id", "food_name", "carbohydrates_g"]
    ].sort_values("carbohydrates_g")
)

print("Number of negative values:", len(negative_carbs))
print("Minimum:", negative_carbs["carbohydrates_g"].min())
print("Maximum:", negative_carbs["carbohydrates_g"].max())
```

```python
negative_ids = set(negative_carbs["fdc_id"])

for food in foods:
    if food["fdcId"] in negative_ids:
        for item in food.get("foodNutrients", []):
            nutrient = item.get("nutrient", {})

            if nutrient.get("id") == 1005:
                print({
                    "fdc_id": food["fdcId"],
                    "food_name": food["description"],
                    "nutrient": nutrient.get("name"),
                    "amount": item.get("amount"),
                    "unit": nutrient.get("unitName"),
                    "data_points": item.get("dataPoints"),
                    "derivation": item.get(
                        "foodNutrientDerivation", {}
                    ).get("description")
                })
```

```python
# Work on a copy, preserving the original extracted dataframe
nutrition_clean_df = nutrition_clean_df.copy()

negative_mask = nutrition_clean_df["carbohydrates_g"] < 0

print("Negative values before cleaning:", negative_mask.sum())

# Replace invalid negative calculated carbohydrate values with NaN
nutrition_clean_df.loc[
    negative_mask, "carbohydrates_g"
] = float("nan")

print(
    "Negative values after cleaning:",
    (nutrition_clean_df["carbohydrates_g"] < 0).sum()
)

print(
    "Missing carbohydrate values after cleaning:",
    nutrition_clean_df["carbohydrates_g"].isna().sum()
)
```

```python
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

output_path = PROCESSED_DIR / "usda_foundation_nutrition.csv"

nutrition_clean_df.to_csv(output_path, index=False)

print("USDA dataset saved!")
print("File:", output_path)
print("Rows:", len(nutrition_clean_df))
print("Columns:", len(nutrition_clean_df.columns))
```
